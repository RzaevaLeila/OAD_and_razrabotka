**Задание 4. Основные статистические показатели**

```python
print('Среднее popularity:', round(df['popularity'].mean(), 2))
print('Медиана popularity:', df['popularity'].median())
```

Нашла среднее и медиану для popularity

```python
print('Минимум:', df['popularity'].min())
print('Максимум:', df['popularity'].max())
print('Стандартное отклонение:', round(df['popularity'].std(), 2))
df['popularity'].describe()
```

Нашла минимум, максимум и стандартное отклонение для popularity. Минимум = 0, максимум = 90, std ≈ 15,95.

**Задание 5. Мода**

```python
print('Мода для genre:', df['genre'].mode().values)
print(df['genre'].value_counts().head())
```

Самый частый жанр — R&B. Дальше идут Pop, Hip-Hop, Latin, Rock

```python
print('Мода для release_year:', df['release_year'].mode().values)
print(df['release_year'].value_counts().sort_index())
```

Чаще всего встречается 2025 год

**Задание 7. Гистограмма**

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, color='pink', edgecolor='black')
plt.title('Распределение популярности треков')
plt.xlabel('Popularity')
plt.ylabel('Количество треков')
plt.grid(axis='y', alpha=0.3)
plt.xticks(range(0, 100, 10))
plt.show()
```

Построила гистограмму. Основная часть треков имеет popularity 10-30. Редко встречаются значения 60–90

**1. Что такое DataFrame?**
