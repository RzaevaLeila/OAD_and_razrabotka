**Задание 4**

```
df_clean = df.dropna(subset=["Year"])
```

```
print("Размер исходного датасета:", df.shape)
print("Размер очищенного датасета:", df_clean.shape)
```

```
print("\nПропуски в df_clean:")
print(df_clean.isnull().sum())
```

**Ответы на вопросы:**

В исходном датасете: 16 600 строк.

После удаления пропусков Year: 16 329 строк.

Пропуски в Year мешают, потому что при группировке по годам строки с NaN либо игнорируются, либо образуют отдельную «пустую» группу, искажая статистику. Нельзя корректно построить динамику продаж по годам, если часть игр не имеет года выпуска.

**Задание 11**

```
corr_na_global = df_clean[["NA_Sales", "Global_Sales"]].corr()
print(corr_na_global)
```

**Ответы:**

Коэффициент корреляции положительный (0.941).

Он очень близок к 1.

Да, можно говорить о выраженной положительной взаимосвязи: чем выше продажи в Северной Америке, тем, как правило, выше мировые продажи.

**Задание 12**

```
plt.figure(figsize=(8, 5))
plt.scatter(df_clean["NA_Sales"], df_clean["Global_Sales"], alpha=0.5, color='purple')
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.xlabel("Продажи в Северной Америке, млн")
plt.ylabel("Мировые продажи, млн")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

**Ответы:**

Одна точка — это одна игра из датасета; её координаты — продажи этой игры в Северной Америке (x) и мировые продажи (y).

Восходящая тенденция наблюдается — точки вытянуты вдоль диагонали из левого нижнего угла в правый верхний.

График соответствует результату корреляционного анализа: сильная положительная связь (r ≈ 0.94).
